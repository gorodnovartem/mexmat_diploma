```python
import numpy as np

# Параметры задачи
#примечание: лямбда с волной в модели соответствует параметру c в коде.
p = 3
c = 1.1
u = 3
x_thresh = 1

# Параметры симуляции
N_max = 500
num_trials = 10000

E_M = np.array([
    c * np.exp(-u) * (1 + (u+1)/k)
    for k in range(1, N_max+1)
])

def sample_Z(k):
    """
    Сэмплируем Z_k по данному F(x)=1 - exp(-x)*(1 + x/k):
    """
    w = (k-1) / k
    if np.random.rand() < w:
        return np.random.exponential(scale=1.0)
    else:
        return np.random.gamma(shape=2.0, scale=1.0)

hits = 0
for _ in range(num_trials):
    S = 0.0
    supS = -np.inf

    for k in range(1, N_max+1):
        # 1) сэмплируем Z_k
        Zk = sample_Z(k)

        # 2) «усекаем» сверху на u
        min_Zk = min(Zk, u)

        # 3) сэмплируем θ_k ~ U[0.2, 2]
        theta_k = np.random.uniform(0.2, 2.2)

        # 4) считаем ξ_k и обновляем сумму
        xi = min_Zk - p*theta_k + E_M[k-1]
        S += xi

        # 5) обновляем супремум
        if S > supS:
            supS = S

        # 6) если превысили порог, фиксируем попадание и выходим
        if supS > x_thresh:
            hits += 1
            break

# Итоговая оценка вероятности
prob = hits / num_trials
print(f"P(sup_k S_k > {x_thresh}) ≈ {prob:.4f}")
#повторить для x in [1;3] с шагом 0.2
```
